```python
import sys
import os
import re
import json
from dataclasses import dataclass, asdict

# Map the path explicitly so the cell can see your local modules
sys.path.append(r"C:\TheFleetOS\02_BRAIN")
from ollama_bridge import OllamaBridge

@dataclass
class RouterResult:
    target_persona: str
    routing_domain: str
    aperture_floor: float
    system_override: str
    payload: str
    priority: str
    memory_scope: str
    stream_mode: bool

    def to_dict(self) -> dict:
        return asdict(self)

class OllamaFleetRouter:
    def __init__(self):
        self.registry = {
            "system_kernel": {
                "routing_domain": "infrastructure",
                "aperture_floor": 0.20,
                "system_override": "You are the automated system kernel.",
                "priority": "high",
                "memory_scope": "local",
                "stream_mode": False,
            },
            "narrative_engine": {
                "routing_domain": "creative",
                "aperture_floor": 0.45,
                "system_override": "You are the narrative story engine.",
                "priority": "normal",
                "memory_scope": "global",
                "stream_mode": False,
            },
            "default_assistant": {
                "routing_domain": "general",
                "aperture_floor": 0.10,
                "system_override": "You are a local context-managed assistant.",
                "priority": "normal",
                "memory_scope": "local",
                "stream_mode": False,
            }
        }
        self.routing_rules = [
            (re.compile(r"(kernel|system|override|infrastructure)", re.IGNORECASE), "system_kernel"),
            (re.compile(r"(narrative|story|continuity|ship|terraformed|planet)", re.IGNORECASE), "narrative_engine")
        ]

    def classify_intent(self, prompt: str) -> str:
        for pattern, persona_key in self.routing_rules:
            if pattern.search(prompt):
                return persona_key
        return "default_assistant"

    def route_request(self, raw_prompt: str) -> dict:
        persona_key = self.classify_intent(raw_prompt)
        config = self.registry.get(persona_key, self.registry["default_assistant"])
        return RouterResult(
            target_persona=persona_key,
            routing_domain=config["routing_domain"],
            aperture_floor=float(config["aperture_floor"]),
            system_override=str(config["system_override"]),
            payload=raw_prompt,
            priority=str(config["priority"]),
            memory_scope=str(config["memory_scope"]),
            stream_mode=bool(config["stream_mode"])
        ).to_dict()

# --- RUN THE PIPELINE HANDSHAKE ---
router = OllamaFleetRouter()

# Explicitly target your local llama3 brain
from ollama_bridge import OllamaContextBridge

test_prompt = "Verify continuity parameters for the automated terraforming hulls."
envelope_data = router.route_request(test_prompt)

print("--- [STAGE 1] GENERATING ROUTING ENVELOPE ---")
print(json.dumps(envelope_data, indent=2))

print("\n--- [STAGE 2] TRANSMITTING TO OLLAMA RUNTIME ---")
try:
    response = bridge.generate(
        prompt=envelope_data["payload"],
        system_prompt=envelope_data["system_override"]
    )
    print("\n--- [STAGE 3] LIVE RESPONSE RECEIVED ---")
    print(response)
except Exception as e:
    print(f"\n[!] Connection Error: Ensure the Ollama desktop application is active. Details: {e}")
```
